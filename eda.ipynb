# Tratamento valores invalidos

### LotFrontage

Esse feature representa o comprimento da frente do terreno que faz divisa com a rua, geralmente medido em pés lineares. Ou seja, ele indica **quantos pés do terreno estão voltados para a rua**, o que pode influenciar fatores como acessibilidade, visibilidade e valorização do imóvel. No tratamento dessa feature nos usamos a mediana com `Neighborhood` x `LotFrontage` porque os terrenos dentro de um mesmo bairro tendem a apresentar características semelhantes.


```python
# print(df_treino["LotFrontage"].value_counts(dropna=False).to_string())

filtro_nan = df_treino["LotFrontage"].isna()

print(df_treino[filtro_nan].value_counts(dropna=False))

df_treino["LotFrontage"] = df_treino.groupby("Neighborhood")["LotFrontage"].transform(
    lambda x: x.fillna(x.median())
)

print(df_treino["LotFrontage"].value_counts(dropna=False).head(5))
# print(df_treino.groupby("Neighborhood")["LotFrontage"].describe())

# df_treino["LotFrontage"] = df_treino.groupby("Neighborhood")["LotFrontage"].transform(
#     lambda x: x.fillna(x.median())
# )
```

### Alley
Se o valor dessa feature for invalido, isso signica que o imovel não possui beco

```python
print(df_treino["Alley"].value_counts(dropna=False).to_string())

# filtro_nan = df_treino["Alley"].isna()

# print(df_treino[filtro_nan].value_counts(dropna=False))

df_treino["Alley"] = df_treino["Alley"].fillna("NA")

print(df_treino["Alley"].value_counts(dropna=False).to_string())
```

### MasVnrArea

Essa feature indica a Área de revestimento de alvenaria em pés quadrados. Caso essa area seja nula pode-se deduzir que ela não existe no imóvel

```python
# filtro_nan = df_treino["MasVnrArea"].isna()

# print(df_treino[filtro_nan].value_counts(dropna=False))
print(df_treino["MasVnrArea"].value_counts(dropna=False).head(5))

df_treino["MasVnrArea"] = df_treino["MasVnrArea"].fillna(0.0)

print(df_treino["MasVnrArea"].value_counts(dropna=False).head(5))
```

### MasVnrType


```python
filtro_nan = (df_treino["MasVnrType"].isna()) & (df_treino["MasVnrArea"] != 0.0)

# print(df_treino[filtro_nan].value_counts(dropna=False).to_string())
print(df_treino[["MasVnrType","MasVnrArea"]].value_counts(dropna=False).head(5))

```

#### Caso a area seja igual a zero
Adere o valor `None` para a feature `MasVnrType` caso a area de `MasVnrArea` for 0. Pois nesse caso a alvenaria não existe

```python
df_treino.loc[
    (df_treino["MasVnrType"].isna()) & (df_treino["MasVnrArea"] == 0.0),
    "MasVnrType"
] = "None"

print(df_treino[["MasVnrType","MasVnrArea"]].value_counts(dropna=False).to_string())
```

#### Existem alvenarias com area mas com o tipo de material invalido. Para não perdemos dados optamos por preencher o tipo de material invalido por `Unknown`

```python
print(df_treino.loc[
    (df_treino["MasVnrType"].isna()) & (df_treino["MasVnrArea"] > 0)
].to_string())

df_treino.loc[
    (df_treino["MasVnrType"].isna()) & (df_treino["MasVnrArea"] > 0),
    "MasVnrType"
] = "Unknown"

print(df_treino["MasVnrType"].value_counts().to_string())
```

## IdadeCasa
Representa a idade da casa em anos. Quando mais velha a casa menor sera o preço

Indica quantos anos se passaram após a ultima reforma feita. Quando uma casa foi recém reformada o seu preço tende a subir

## AreaTotal
Indica a area total interna. Quando mais espaçosa a casa mais cara ela tende á ser

## QtdBanheiros
Indica a quantidade de banheiros presentes na casa. Como indicado no grafico quanto maior a quantidade de banheiros maior vai ser o preço

```python
df_treino["QtdBanheiros"] = (
    df_treino["BsmtFullBath"] + (0.5 * df_treino["BsmtHalfBath"]) + df_treino["FullBath"] + (0.5 * df_treino["HalfBath"])
)

sns.barplot(data=df_treino,x="QtdBanheiros",y="SalePrice")
plt.show()
```

# Simetria dos dados

Foi feito grafico de barras com cauda para visualizar as simetria das colunas. Foi escolhido um limiar de 2.0 para filtrar apenas aquelas que apresentam alta assimetria

# Simetria da variavel alvo `SalePrice`

```python
sns.histplot(df_treino["SalePrice"], kde=True)
plt.title("distribuição da SalePrice")
plt.show()
```

Como a distribuição da varivel alvo é uma assimetria positiva. Foi aplicada uma transformação logaritmica para normalizar a distribuição dessa variavel

```python
df_treino["SalePrice"] = np.log1p(
    df_treino["SalePrice"]
)

sns.histplot(df_treino["SalePrice"], kde=True)
plt.title("distribuição da SalePrice (depois da normalização)")
plt.show()
```

Aplicamos a transformação logaritmica apenas nas colunas que não apresentam um grande volume de zeros presentes. Como na `LotFrontage`, `LotArea`

```python
colunas = ["LotFrontage", "LotArea"]

# guarda valores antes da transformação
df_antes = df_treino[colunas].copy()

# transformação
for coluna in colunas:
    df_treino[coluna] = np.log1p(df_treino[coluna])

# cria layout 2 linhas x N colunas
fig, axes = plt.subplots(2, len(colunas), figsize=(12, 8))

for i, coluna in enumerate(colunas):

    # antes
    sns.histplot(df_antes[coluna], kde=True, ax=axes[0, i])
    axes[0, i].set_title(f"{coluna} - Antes")

    # depois
    sns.histplot(df_treino[coluna], kde=True, ax=axes[1, i])
    axes[1, i].set_title(f"{coluna} - Depois")

plt.tight_layout()
plt.show()
```

# Correlação entre os dados

Como visto na matriz de correlação quando o valor dessas features aumentam o valor do preço da casa, tambem, aumenta. Como visto com a relação `OverallQual` x `SalePrice`

```python
colunas_numericas = df_treino.select_dtypes(include="number")

correlacoes = (
    colunas_numericas.corr(method="spearman")["SalePrice"]
    # .drop("SalePrice")   # remove a própria target
    .drop("Id", errors="ignore")
)

top5 = correlacoes.abs().sort_values(ascending=False).head(5)
colunas_top5 = top5.index.tolist()

cols_heatmap = colunas_top5

# plt.figure(figsize=(8,6))
plt.figure()

sns.heatmap(
    df_treino[cols_heatmap].corr(),
    cmap="coolwarm",
    annot=True
)
plt.title("Correlação das Top 5 com SalePrice")

plt.show()
```

# Identificação de outliers
A seleção de features foi feita usando a matriz de correlação usando o metodo de spearman. Foram selecionadas aquelas que possuiam uma correlação maior que 0.5 com a feature alvo

```python
colunas_numericas = df_treino.select_dtypes(include="number")
colunas_correlacionadas = colunas_numericas.corr("spearman")["SalePrice"].sort_values(ascending=False)
colunas_relevantes = colunas_correlacionadas[abs(colunas_correlacionadas) > 0.5].drop("SalePrice")

print("colunas relevantes x correlação")
print(colunas_relevantes)
```
