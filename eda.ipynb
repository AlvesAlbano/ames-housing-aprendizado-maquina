# Tratamento valores invalidos

### LotFrontage

Esse feature representa o comprimento da frente do terreno que faz divisa com a rua, geralmente medido em pés lineares. Ou seja, ele indica **quantos pés do terreno estão voltados para a rua**, o que pode influenciar fatores como acessibilidade, visibilidade e valorização do imóvel. No tratamento dessa feature nos usamos a mediana com `Neighborhood` x `LotFrontage` porque os terrenos dentro de um mesmo bairro tendem a apresentar características semelhantes.


```python
# print(df_treino["LotFrontage"].value_counts(dropna=False).to_string())

filtro_nan = df_treino["LotFrontage"].isna()

print(df_treino[filtro_nan].value_counts(dropna=False))

df_treino["LotFrontage"] = df_treino.groupby("Neighborhood")["LotFrontage"].transform(
    lambda x: x.fillna(x.median())
)

print(df_treino["LotFrontage"].value_counts(dropna=False))
# print(df_treino.groupby("Neighborhood")["LotFrontage"].describe())

# df_treino["LotFrontage"] = df_treino.groupby("Neighborhood")["LotFrontage"].transform(
#     lambda x: x.fillna(x.median())
# )
```

# Identificação de outliers
A seleção de features foi feita usando a matriz de correlação usando o metodo de spearman. Foram selecionadas aquelas que possuiam uma correlação maior que 0.5 com a feature alvo

```python
colunas_numericas = df_treino.select_dtypes(include="number")
colunas_correlacionadas = colunas_numericas.corr("spearman")["SalePrice"].sort_values(ascending=False)
colunas_relevantes = colunas_correlacionadas[abs(colunas_correlacionadas) > 0.5].drop("SalePrice")

print("colunas relevantes x correlação")
print(colunas_relevantes)
```

## Usando um grafico de dispersão + IQR

Foi usado o IQR para calcular os valores extremos inferiores e superiores de cada coluna relevante

```python
def outliers_IQR(df,coluna):
    Q1 = df[coluna].quantile(0.25)
    Q3 = df[coluna].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[coluna] < lower) | (df[coluna] > upper)]

    return outliers

for coluna in colunas_relevantes.index:
    outliers = outliers_IQR(df_treino,coluna)

    sns.scatterplot(data=df_treino,x=coluna,y="SalePrice")
    sns.scatterplot(data=outliers,x=coluna,y="SalePrice",marker="s",color="firebrick")

    plt.title(f"{coluna} x SalePrice")

    plt.show()
```

Já que identificamos os outliers iremos remove-los

```python
def remover_outliers_IQR(df,coluna):
    Q1 = df[coluna].quantile(0.25)
    Q3 = df[coluna].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df_sem_otliers = df[(df[coluna] >= lower) & (df[coluna] <= upper)]

    return df_sem_otliers

linhas_antes = df_treino.shape[0]
print(f"antes da remoção dos outliers: {linhas_antes} linhas")
for coluna in colunas_relevantes.index:

    antes = df_treino.shape[0]

    df_limpo = remover_outliers_IQR(
        df_treino,
        coluna
    )

    depois = df_limpo.shape[0]

    removidos = antes - depois

    df_treino = df_limpo

    print(f'\n{coluna}')
    print(f'Antes: {antes}')
    print(f'Depois: {depois}')
    print(f'Removidos: {removidos}')
linhas_depois = df_treino.shape[0]
print(f"depois da remoção dos outliers: {linhas_depois} linhas")
print(f"{linhas_antes - linhas_depois} linhas removidas")
```
